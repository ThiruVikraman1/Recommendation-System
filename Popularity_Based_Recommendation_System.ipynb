```python
# importing necessary libraries
import pandas as pd
import numpy as np
```

```python
# Loading the data
r_cols =['user_id','movie_id','rating','unic_timestamp']
ratings= pd.read_csv('ml-100K/u.data',sep='\t',names=r_cols,encoding='latin-1')
ratings
```

```python
#Loading the movie data
i_cols=['movie id', 'movie title' ,'release date','video release date', 'IMDb URL', 'unknown', 'Action', 'Adventure',
    'Animation', 'Children\'s', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
    'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
items = pd.read_csv('ml-100k/u.item',sep='|',names = i_cols, encoding='latin-1')
items.head(3)
```

```python
#select only needed item
movie_content = items[['movie id','movie title','Action', 'Adventure',
    'Animation', 'Children\'s', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
    'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']]
movie_content
```

```python
#Here we are counting and calculating mean 
movie_stats = ratings.groupby('movie_id')['rating'].agg(['count', 'mean']).reset_index()
movie_stats.head()
```

```python
movie_content = movie_content.rename(columns={'movie id': 'movie_id'})

#Merge movie stats and movie content dataframes on movie_id
movie_data = pd.merge(movie_stats, movie_content, on='movie_id')
movie_data.head()
```

```python
genres = movie_content.columns[2:]
genres
```

```python
# Checking for 1 gener
top5_movies_by_genre = {}

for genre in genres:
    genre_movies = movie_data[movie_data[genre] == 1]
    
    top5 = genre_movies.sort_values(
        by=['count', 'mean'], 
        ascending=False
    ).head(5)[['movie_id', 'movie title', 'count', 'mean']]
    
    top5_movies_by_genre[genre] = top5

top5_movies_by_genre['Adventure']
```

```python
def recommend_by_genre(user_genre=None):

    if user_genre and user_genre in top5_movies_by_genre:
        print(f"\nTop 5 popular movies in {user_genre}:\n")
        return top5_movies_by_genre[user_genre]

    else:
        print("\nTop 5 popular movies from each genre:")
        
        for genre, df in top5_movies_by_genre.items():
            print(genre)
            display(df)
```

```python
recommend_by_genre()
```
